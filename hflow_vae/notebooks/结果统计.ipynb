```python
from pathlib import Path
import pandas as pd


root = Path().cwd().parent
RESULT_DIR = root / "results" / "per_fold"

def summarize_dataset(dataset_name: str):
    files = sorted(RESULT_DIR.glob(f"{dataset_name}__*.parquet"))

    if len(files) == 0:
        raise FileNotFoundError(f"No result files found for {dataset_name}")

    df = pd.concat(
        [pd.read_parquet(f) for f in files],
        ignore_index=True,
    )

    metrics = ["Recall", "F1_score", "G_mean"]

    summary = (
        df.groupby(["Method", "Classifier"])[metrics]
        .agg(["mean", "std"])
        .round(4)
    )

    return summary


if __name__ == "__main__":
    summary = summarize_dataset("breast_cancer_coimbra")
    print(summary)
```
