### Skewness统计

```python
from scipy.stats import skew

def compute_skewness_table(df, numerical_cols):
    """
    计算每个数值列的偏度,并标记中度、重度偏斜
    """
    results = []
    for col in numerical_cols:
        data = df[col].dropna()
        s = skew(data)
        results.append({
            'column': col,
            'skewness': round(s, 3),
            'abs_skew': round(abs(s), 3),
            'category': 'severe (|skew|>1)' if abs(s)>2 
            else 'moderate (|skew|>1)' if abs(s) > 1 
            else 'approx. symmetric'
        })
    return pd.DataFrame(results)
```

### KDE密度图 + 多峰检测

```python
from scipy.stats import gaussian_kde
from scipy.signal import find_peaks
import matplotlib.pyplot as plt


def plot_kde_with_peaks(real_data, synthetic_data=None, col_name='', 
                          prominence_ratio=0.05, ax=None, pad_ratio=0.15):
    real_data = np.asarray(real_data).flatten()
    real_data = real_data[~np.isnan(real_data)]

    x_min, x_max = real_data.min(), real_data.max()
    x_range = x_max - x_min
    # 关键修改：向左右各扩展pad_ratio比例的范围,让边界峰变成网格内部点
    grid_min = x_min - pad_ratio * x_range
    grid_max = x_max + pad_ratio * x_range
    x_grid = np.linspace(grid_min, grid_max, 500)

    kde_real = gaussian_kde(real_data)
    y_real = kde_real(x_grid)

    if ax is None:
        fig, ax = plt.subplots(figsize=(6, 4))
    ax.plot(x_grid, y_real, label='Real', color='black')
    ax.fill_between(x_grid, y_real, alpha=0.15, color='black')

    if synthetic_data is not None:
        synthetic_data = np.asarray(synthetic_data).flatten()
        synthetic_data = synthetic_data[~np.isnan(synthetic_data)]
        kde_syn = gaussian_kde(synthetic_data)
        y_syn = kde_syn(x_grid)
        ax.plot(x_grid, y_syn, label='Synthetic', color='orange', linestyle='--')

    peaks, properties = find_peaks(y_real, prominence=prominence_ratio * y_real.max())
    ax.scatter(x_grid[peaks], y_real[peaks], color='red', zorder=5,
               label=f'{len(peaks)} peak(s)')

    # 同时画出实际数据范围的边界线,方便核对"检测到的峰是否落在真实数据范围内"
    ax.axvline(x_min, color='gray', linestyle=':', alpha=0.5)
    ax.axvline(x_max, color='gray', linestyle=':', alpha=0.5)

    ax.set_title(col_name)
    ax.set_xlabel(col_name)
    ax.set_ylabel('Density')
    ax.legend()
    return len(peaks), x_grid[peaks].tolist()
```
